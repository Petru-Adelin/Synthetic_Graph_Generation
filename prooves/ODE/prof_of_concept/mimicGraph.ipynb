```python
import networkx as ntx 
import numpy as np
import matplotlib.pyplot as plt
import math 
```

```python
origina_graph = ntx.Graph()
P = .01
N, E = [i for i in range(25)], []

# a list of all the edges that are not in the graph
exclusive_E = []

# generate the random graph
for x in N:
    for y in range(x+1, len(N)):
        if x == y:
            continue
        if np.random.choice([True, False], p=[P, 1-P]):
            E = [*E, (x, y)]
        else:
            exclusive_E = [*exclusive_E, (x, y)]


origina_graph.add_nodes_from(N)
origina_graph.add_edges_from(E)

# display 
ntx.draw_circular(origina_graph)
plt.show()
print(len(E), len(exclusive_E))
print(exclusive_E)
```

```python
# ODE that describes the graph 
copy_graph = origina_graph.copy()

for t in range(10, 150, 10):
    additional_E = []
    P = 0.01 * (math.e ** (t * 0.02))
    for edge in exclusive_E:
        if np.random.choice([True, False], p=[P, 1-P]):
            additional_E = [*additional_E, edge]
            exclusive_E.remove(edge)
    copy_graph.add_edges_from(additional_E)
    ntx.draw_circular(copy_graph)
    plt.show()
    print(len(exclusive_E))
```
