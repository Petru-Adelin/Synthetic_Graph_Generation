```python
# global imports
import networkx as ntx
import matplotlib.pyplot as plt
import numpy as np
import math 
```

```python
# consider the next scenario we have a temporal graph that is given in  the initial state 
# init_state = (N, E); where N - number of nodes / E - number of edges 
# we will use a basic generative model of a random graph (Erdos-Renyi) model 
# The Erdos model chooses with probability P for each pair of nodes if there will be an edge 
g = ntx.Graph()
P = .01
N, E = [], []
# adding the nodes
for i in range(1, 26):
    N = [*N, i]
g.add_nodes_from(N)
# adding the edges
for x in N:
    for y in N:
        # jump over the loops in the graph
        if x == y:
            continue
        is_there_edge = np.random.choice([True, False], p=[P, 1-P])
        if is_there_edge:
            E = [*E, (x, y)]

print(E, N)

g.add_edges_from(E)


ntx.draw_circular(g)
plt.show()
```

```python
# Lets presume a basic example for the proof
# Lets imagine the initial state graph given above has a rate of change of 2% in the probability 
# In a Social Network that would literally describe the tendecy of people creating stronger groups over time 

# So for this example the attached differential equation that describe the system will be 
# dy/dt = 0.02 * y 
# given the calculation we get: y = C * (e ^ (0.02*t)) where C must be resolved based on the initial state
# given y(0) = 0.01 => C = 0.01 => the final result is: y = 0.01 * (e ^ (0.02 * t))

# now lets see how the network changes in time
# I will take around 15 frames in different time periods each at 10 time units one after another


for t in range(10, 151, 10):
    
    new_P = 0.01 * (math.e ** (0.02 * t))
    print(new_P)
    # recreate the graph
    new_g = ntx.Graph() 
    new_g.add_nodes_from(N)
    new_E = []
    for x in N:
        for y in N:
            # jump over the loops in the graph
            if x == y:
                continue
            is_there_edge = np.random.choice([True, False], p=[new_P, 1-new_P])
            if is_there_edge:
                new_E = [*new_E, (x, y)]
    
    print(new_E)
    new_g.add_edges_from(new_E)

    ntx.draw_circular(new_g)
    plt.show()
    
    
```
